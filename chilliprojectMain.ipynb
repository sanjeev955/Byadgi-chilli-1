Convert HEIC to JPG

```python
from pillow_heif import register_heif_opener
from PIL import Image
import os

register_heif_opener()

folders = [
    '/content/drive/MyDrive/ChilliDataset/train',
    '/content/drive/MyDrive/ChilliDataset/test'
]

for input_folder in folders:
    for root, dirs, files in os.walk(input_folder):
        for file in files:
            if file.endswith(".heic") or file.endswith(".HEIC"):
                heic_path = os.path.join(root, file)
                jpg_path = heic_path.replace(".heic", ".jpg").replace(".HEIC", ".jpg")

                try:
                    image = Image.open(heic_path)
                    image.convert("RGB").save(jpg_path, "JPEG")
                    print("Converted:", file)
                except:
                    print("Error:", file)
```

Import Libraries

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import numpy as np
import cv2
import matplotlib.pyplot as plt
from PIL import ImageFile
ImageFile.LOAD_TRUNCATED_IMAGES = True
```

Data Preprocessing

```python
train_datagen = ImageDataGenerator(
        rescale=1./255,
        shear_range=0.2,
        zoom_range=0.2,
        horizontal_flip=True)
```

Load Training Dataset

```python
training_set = train_datagen.flow_from_directory(
        '/content/drive/MyDrive/ChilliDataset/train',
        target_size=(64,64),
        batch_size=64,
        class_mode='categorical')
```

Load Test Dataset

```python
test_datagen = ImageDataGenerator(rescale=1./255)

test_set = test_datagen.flow_from_directory(
        '/content/drive/MyDrive/ChilliDataset/test',
        target_size=(64,64),
        batch_size=64,
        class_mode='categorical')
```

Build CNN Model

```python
cnn = tf.keras.models.Sequential()
```

First Convolution Layer

```python
cnn.add(tf.keras.layers.Conv2D(
    filters=32,
    kernel_size=3,
    activation='relu',
    input_shape=[64,64,3]
))
```

Pooling Layer

```python
cnn.add(tf.keras.layers.MaxPool2D(
    pool_size=2,
    strides=2
))
```

Second Convolution Layer

```python
cnn.add(tf.keras.layers.Conv2D(
    filters=32,
    kernel_size=3,
    activation='relu'
))
```

Second Pooling

```python
cnn.add(tf.keras.layers.MaxPool2D(
    pool_size=2,
    strides=2
))
```

Flatten Layer

```python
cnn.add(tf.keras.layers.Flatten())
```

Dense Layer

```python
cnn.add(tf.keras.layers.Dense(
    units=128,
    activation='relu'
))
```

Output Layer

```python
cnn.add(tf.keras.layers.Dense(
    units=4,
    activation='softmax'
))
```

Compile Model

```python
cnn.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

Train Model

```python
cnn.fit(
    x=training_set,
    validation_data=test_set,
    epochs=15
)

cnn.save('/content/chilli_model.h5')

print("Model saved")
```

Color Detection (Red Quality)

```python
def detect_color(image_path):

    img = cv2.imread(image_path)

    hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

    lower_red = np.array([0,120,70])
    upper_red = np.array([10,255,255])

    mask = cv2.inRange(hsv, lower_red, upper_red)

    red_pixels = np.sum(mask > 0)

    print("Red Pixel Count:", red_pixels)
```

Chilli Length Detection

```python
def detect_length(image_path):

    img = cv2.imread(image_path)

    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    edges = cv2.Canny(gray,100,200)

    contours,_ = cv2.findContours(edges,cv2.RETR_EXTERNAL,cv2.CHAIN_APPROX_SIMPLE)

    c = max(contours,key=cv2.contourArea)

    x,y,w,h = cv2.boundingRect(c)

    length = max(w,h)

    print("Chilli Length:", length)
```

Wrinkle Detection

```python
def detect_wrinkles(image_path):

    img = cv2.imread(image_path)

    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    edges = cv2.Canny(gray,50,150)

    wrinkle_score = np.sum(edges)

    print("Wrinkle Score:", wrinkle_score)
```

Cell 2 (API)

```python
from flask import Flask, request, jsonify
from flask_ngrok import run_with_ngrok
import tensorflow as tf
import numpy as np
from PIL import Image

app = Flask(__name__)
run_with_ngrok(app)

model = tf.keras.models.load_model('/content/chilli_model.h5')

classes = ['DHQ','DLQ','KHQ','KLQ']

@app.route('/predict', methods=['POST'])
def predict():
    file = request.files['file']

    img = Image.open(file).resize((64,64))
    img = np.array(img)/255.0
    img = np.expand_dims(img, axis=0)

    prediction = model.predict(img)
    result = classes[np.argmax(prediction)]

    return jsonify({'prediction': result})

app.run()
```
